```python
import numpy as np
from sklearn.preprocessing import MinMaxScaler   # M aur S capital honi chahiye
from random import randint
import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense
from tensorflow.keras.optimizers import Adam

print("Import successful!")
```

```python
train_sample=[]
train_label=[]
```

```python
for i in range(1000):
    younger_ages=randint(13,64)
    train_sample.append(younger_ages)
    train_label.append(0)

    older_ages=randint(65,100)
    train_sample.append(older_ages)
    train_label.append(1)
```

```python
train_sample,train_label
```

```python
train_sample=np.array(train_sample)
train_label=np.array(train_label)
```

```python
scaler=MinMaxScaler(feature_range=(0,1))
scaler_train_sample=scaler.fit_transform(train_sample.reshape(-1,1))
```

```python
scaler_train_sample
```

```python
model=Sequential([
    Dense(16,input_dim=1,activation='relu'),
    Dense(32,activation='relu'),
    Dense(2,activation='softmax')])
```

```python
model.summary()
```

```python
model.compile(Adam(learning_rate=0.001),loss="sparse_categorical_crossentropy",metrics=['accuracy'])
```

```python
model.fit(train_sample,train_label,batch_size=10,epochs=10)
```

```python
test_sample=[]
test_label=[]
```

```python
for i in range(500):
    younger_ages=randint(13,64)
    test_sample.append(younger_ages)
    test_label.append(0)

    older_ages=randint(65,100)
    test_sample.append(older_ages)
    test_label.append(1)
```

```python
test_sample=np.array(test_sample)
test_label=np.array(test_label)
```

```python
predictions = model.predict(test_sample, batch_size=10)
predicted_classes = np.argmax(predictions, axis=1)

print(predicted_classes)
```

```python
from sklearn.metrics import confusion_matrix
```

```python
predicted_values=confusion_matrix(test_label,predicted_classes)
```
